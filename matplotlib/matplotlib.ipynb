```python
batsman=pd.read_csv("C:/Users/Dell/Downloads/sharma-kohli.csv")
batsman
```

```python


plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
#plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#label title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title("rohit and virat camparision")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
#for giving color
plt.plot(batsman['index'],batsman['V Kohli'],color="black")
plt.plot(batsman['index'],batsman['RG Sharma'],color="red")
plt.title("rohit and virat camparision")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
#for line width
plt.plot(batsman['index'],batsman['V Kohli'],color="black",linestyle="dashed",marker="+",markersize=10)#dashed_doted
plt.plot(batsman['index'],batsman['RG Sharma'],color="red",linestyle="dotted",linewidth=4)
plt.title("rohit and virat camparision")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
#legend
plt.plot(batsman['index'],batsman['V Kohli'],color="black",linestyle="dashed",marker="+",markersize=10,label="virat")#dashed_doted
plt.plot(batsman['index'],batsman['RG Sharma'],color="red",linestyle="dotted",linewidth=4,label="rohit")
plt.title("rohit and virat camparision")
plt.xlabel("season")
plt.ylabel("runs scored")
plt.legend(loc="upper right")
```

```python
#limiting axes 
price=[48000,54000,57000,49000,45000,4500000]#it does no show all the progress it directly show larger number progress  
year=[2015,2016,2017,2018,2020,2025]
plt.plot(year,price)
plt.ylim(0,150000) #therefor we use this 
```

```python
#grid:It give more readablity
plt.plot(batsman['index'],batsman['V Kohli'],color="black",linestyle="dashed",marker="+",markersize=10,label="virat")#dashed_doted
plt.plot(batsman['index'],batsman['RG Sharma'],color="red",linestyle="dotted",linewidth=4,label="rohit")
plt.title("rohit and virat camparision")
plt.xlabel("season")
plt.ylabel("runs scored")
plt.legend(loc="upper right")
plt.grid()
```

```python
#show()-used for showing the graph
plt.plot(batsman['index'],batsman['V Kohli'],color="black",linestyle="dashed",marker="+",markersize=10,label="virat")#dashed_doted
plt.plot(batsman['index'],batsman['RG Sharma'],color="red",linestyle="dotted",linewidth=4,label="rohit")
plt.title("rohit and virat camparision")
plt.xlabel("season")
plt.ylabel("runs scored")
plt.legend(loc="upper right")
plt.grid()
plt.show()
```
