```python
#converting to datetime datatype
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
#extract year
date['date_year']=date['date'].dt.year
date.head()
```

```python
#extract month
date['date_month']=date['date'].dt.month
date.sample(5)
```

```python
#extract month name
date['month_name']=date['date'].dt.month_name()
date.head()
```

```python
#extract days
date['date_days']=date['date'].dt.day
date.head()
```

```python
#day of week
date['date_dow']=date['date'].dt.dayofweek
date.head()
```

```python
#day of week name
date['dofn']=date['date'].dt.day_name()
date.head()
```

```python
#is weekend?
date['date_is_weekend']=np.where(date['dofn'].isin(['Sunday','Saturday']),1,0)
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
#extract week of year
date['date_week']=date['date'].dt.isocalendar().week
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['date_quarter']=date['date'].dt.quarter
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['semister']=np.where(date['date_quarter'].isin([1,2]),1,2)
date.drop(columns=['product_id','city_id','orders']).head()
```
