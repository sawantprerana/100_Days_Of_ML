```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
webpage=requests.get('https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav',headers=headers).text
```

```python
soup=BeautifulSoup(webpage,'lxml')#lxml because of this we can traverse easily in html code
```

print(soup.prettify())#it is used to format the html code

# Names Of All company

```python
for i in soup.findAll('h2'):
    print(i.text.strip())
```

```python
len(soup.findAll('h2'))
```

# For Finding Rating

```python
for i in soup.findAll('div',class_="rating_text"):
  print(i.text.strip())
     
```

```python
len(soup.findAll('div',class_="rating_text"))
```

```python
company=soup.find_all('div',class_='companyCardWrapper')
```

```python
len(company)
```

```python
name = []
rating = []
reviews = []
salaries = []
interviews = []
jobs = []

for i in company:
    name.append(i.find('h2').text.strip())
    rating.append(i.find('div', class_='rating_text').text.strip())
    
    
    actions = i.find_all('span', class_='companyCardWrapper__ActionCount')
    
    reviews.append(actions[0].text.strip())
    salaries.append(actions[1].text.strip())
    interviews.append(actions[2].text.strip())
    jobs.append(actions[3].text.strip())

df = pd.DataFrame({
    'name': name,
    'rating': rating,
    'reviews': reviews,
    'salaries': salaries,
    'interviews': interviews,
    'jobs': jobs
})


```
