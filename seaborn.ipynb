```python
#scatterplot
#usually done on numrical column
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px  #for dataset that we can direcly use
```

```python
tips=sns.load_dataset("tips")
tips
```

```python
#It is axis level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#relplot->figure level->It is in square shape
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
#lineplot
gap=px.data.gapminder()
df=gap[gap['country']=='India']
df
```

```python
sns.lineplot(data=df,x='year',y='lifeExp')
```

```python
sns.relplot(data=df,x='year',y='lifeExp',kind='line',hue='')
```

```python
gap.sample(10)
```

```python
temp_df=gap[gap['country'].isin(['India','Pakistan','China'])]
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',) # it is fig level function so that legend is outof axis
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp',hue='country',style='continent')#here we can see that that legend is in the axis bacause it is axis level level function 
```

```python
#facet plot->figure level function->work with relplot
#it will not work with axis level function like scatterplot function and lineplot function
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex',row='smoker')
```

```python
#col wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=4)
```

```python
#distribution plot
#used for univariate analysis
#it is used to find disrtibution
#are there outliers
#plots->histplot,kdeplot,rugplot

#fig level function->displot
#axes level function->histplot,rugplot,kdeplot
```

```python
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,kind='hist',x='total_bill',bins=10)
```

```python
#it  can be also used on categorical column
#it is also  known as countplot
sns.displot(data=tips,x='day',kind='hist')
```

```python
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='sex',element='step') #if the data gets overlap and we want to see it clearly then we use element='step'
```

```python
#kdeplot(Kernel Density Estimation)
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,kind='kde',x='total_bill',hue='sex',fill=True)
```
